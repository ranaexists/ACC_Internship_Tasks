# Project 3: Scraping Data into Pandas DataFrame and Exporting to CSV

**Objective:** Extract multiple elements from a webpage, structure them using a `pandas` DataFrame, and save the final dataset as a CSV file for analysis.

```python
# Importing required libraries
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "http://quotes.toscrape.com/"
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')

# Finding all quotes and authors on the page
quotes = soup.find_all('span', class_='text')
authors = soup.find_all('small', class_='author')

# Creating empty lists to store our extracted data
quote_list = []
author_list = []

# Looping through all extracted elements and adding text to our lists
for i in range(len(quotes)):
    quote_list.append(quotes[i].text)
    author_list.append(authors[i].text)

# Structuring the data into a Pandas DataFrame
df = pd.DataFrame({
    'Quote': quote_list,
    'Author': author_list
})

# Saving the DataFrame to a CSV file in the same folder
df.to_csv('scraped_quotes.csv', index=False)

print("Data successfully saved to 'scraped_quotes.csv'!")

# Displaying the first 5 rows to verify
print(df.head())
```
