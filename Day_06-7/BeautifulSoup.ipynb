# Project 2: Static Web Scraping with BeautifulSoup

**Objective:** Extract HTML content from a website using `requests` and parse specific elements (Quotes and Authors) using the `BeautifulSoup` library.

```python
# Import required libraries
import requests
from bs4 import BeautifulSoup

# The practice website we want to scrape
url = "http://quotes.toscrape.com/"

# Get the webpage content
response = requests.get(url)

# Convert the raw HTML into a BeautifulSoup object to easily search through it
soup = BeautifulSoup(response.text, 'html.parser')

# Find all HTML elements that contain the quotes and authors
# (On this website, quotes are in <span class="text"> and authors in <small class="author">)
quotes = soup.find_all('span', class_='text')
authors = soup.find_all('small', class_='author')

# Loop through the first 3 results and print them cleanly
print("--- Top 3 Quotes ---")
for i in range(3):
    # .text removes the HTML tags and gives us just the plain text
    quote_text = quotes[i].text
    author_name = authors[i].text
    
    print(f"Quote: {quote_text}")
    print(f"Author: {author_name}")
    print("-" * 40)
```
