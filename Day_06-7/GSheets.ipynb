# Project 5: No-Code Web Scraping using Google Sheets

**Objective:** Understand and demonstrate how to quickly scrape tabular data from web pages without writing Python scripts, using Google Sheets built-in functions.

### Implementation Steps:
Instead of writing a Python script, I used Google Sheets for quick data extraction. This is a highly efficient "No-Code" approach for simple tabular data.

**Target:** Scrape the table of "Largest companies by revenue" from Wikipedia.

**The Formula Used in Google Sheets:**
`=IMPORTHTML("https://en.wikipedia.org/wiki/List_of_largest_companies_by_revenue", "table", 1)`

**Result:**
The formula successfully extracted the first table from the Wikipedia page and populated it directly into the spreadsheet rows and columns, saving time compared to a full Python BeautifulSoup setup.
