# Project 1: Data Extraction Using API

**Objective:** Fetch user data from a public API using the `requests` library and extract specific fields (Name and Email).

```python
# Importing the requests library to communicate with the web
import requests

# This is the public API endpoint we are getting data from
url = "https://jsonplaceholder.typicode.com/users"

# Sending a GET request to the URL
response = requests.get(url)

# Checking if the connection was successful (200 means everything is OK)
if response.status_code == 200:
    
    # Converting the raw response into JSON format so Python can read it easily
    users_data = response.json()
    
    # Looping through just the first 3 users in our data
    for user in users_data[:3]:
        
        # Printing out the specific information we need
        print(f"Name: {user['name']} | Email: {user['email']}")

else:
    # If the request fails, show the error code
    print(f"Failed to fetch data. Status code: {response.status_code}")
```
