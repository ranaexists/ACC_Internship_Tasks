# Project 4: Pagination Web Scraping

**Objective:** Use a loop to navigate through multiple pages (pagination) on a website, scrape data from each page, and consolidate it into a single Pandas DataFrame.

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

# Empty lists to hold data from all pages
all_quotes = []
all_authors = []

# Loop through pages 1 to 3
for page_num in range(1, 4):
    
    # Constructing the URL dynamically for each page
    url = f"http://quotes.toscrape.com/page/{page_num}/"
    print(f"Scraping page {page_num}...")
    
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')
    
    # Finding quotes and authors for the current page
    quotes = soup.find_all('span', class_='text')
    authors = soup.find_all('small', class_='author')
    
    # Extracting text and adding to our main lists
    for i in range(len(quotes)):
        all_quotes.append(quotes[i].text)
        all_authors.append(authors[i].text)
    
    # Adding a small 1-second delay so we don't overload the server (Good practice)
    time.sleep(1)

# Creating the final DataFrame
df_multi_page = pd.DataFrame({
    'Quote': all_quotes,
    'Author': all_authors
})

print("-" * 30)
print(f"Total quotes scraped: {len(df_multi_page)}")
print(df_multi_page.head())
```
